```python
from sklearn.datasets import fetch_california_housing ## importar funciones para obtener el dataset y cargar los datos de las viviendas
california = fetch_california_housing()
```

```python
## Convertir los datos en una tabla de pandas

import pandas as pd

df_california = pd.DataFrame(
    california.data,
    columns=california.feature_names
) ## contiene la información de las viviendas

df_california["MedHouseVal"] = california.target ## variable valor mediano de las viviendas

df_california.head()
```
